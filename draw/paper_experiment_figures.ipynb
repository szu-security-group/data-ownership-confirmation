## Figure 4 — Performance Evaluation on Real-World Datasets

左图：统一颜色的构建时间柱状图。右图：对数纵轴的分组点图，展示证明生成与增删改耗时。


```python
compact_style()
plt.rcParams.update({'font.family': 'serif',
                     'font.serif': ['Times New Roman', 'DejaVu Serif'],
                     'mathtext.fontset': 'stix'})
datasets = ['coco', 'clevr', 'librispeech']
names = ['MS COCO', 'CLEVR', 'LibriSpeech']

def real_values(op, divisor):
    return [ht_value('real', op, dataset=d) / divisor for d in datasets]

def draw_real_panel(ax, index):
    if index == 0:
        ax.bar(range(3), real_values('build', 1e6), width=.5,
               color='#4C72B0', edgecolor='black', linewidth=.4, alpha=.9)
        ax.set(ylabel='Time cost (ms)', ylim=(0,150), yticks=[0,50,100,150])
    else:
        # Grouped points: horizontal offsets distinguish the operations;
        # data points from different datasets are not connected.
        for offset, op, color, marker, label in zip(
                [-1.5,-.5,.5,1.5], ['prove','insert','delete','modify'],
                ['#4C72B0','#55A868','#DD8452','#C44E52'],
                ['o','^','D','s'], ['Proof Gen','Insert','Delete','Modify']):
            ax.plot(np.arange(3)+offset*.18, real_values(op, 1e3),
                    linestyle='none', marker=marker, markersize=3.2,
                    markerfacecolor='white', markeredgewidth=.7,
                    color=color, label=label)
        ax.set_yscale('log')
        ax.set(ylabel='Time cost (µs)', ylim=(.2,150))
        ax.set_yticks([1,10,100], ['1','10','100'])
        ax.yaxis.set_minor_locator(NullLocator())
        ax.set_xlim(-.55,2.55)
        ax.legend(loc='upper left', fontsize=6, frameon=False,
                  handlelength=1, handletextpad=.4, borderpad=.1, labelspacing=.18)
    ax.spines['top'].set_visible(True)
    ax.spines['right'].set_visible(True)
    ax.set_xticks(range(3), names)
    ax.tick_params(axis='x', labelsize=5.5)
    ax.set_axisbelow(True)
    ax.grid(axis='y', linestyle='--', alpha=.35, linewidth=.4)

# Export at the exact original PDF size: 1.59 x 1.45 inches per panel.
# Original performance_current_a/b.pdf files are not overwritten.
if EXPORT_PDF:
    EXPORT_DIR.mkdir(parents=True, exist_ok=True)
    for index, letter in enumerate('ab'):
        panel = plt.figure(figsize=(1.59,1.45))
        draw_real_panel(panel.add_axes([.245,.18,.735,.795]), index)
        panel.savefig(EXPORT_DIR / f'performance_log_{letter}.pdf')
        plt.close(panel)

# Combined view for browsing in the notebook.
fig = plt.figure(figsize=(3.53,1.72))
axes = [fig.add_axes([.10,.29,.36,.67]),fig.add_axes([.62,.29,.36,.67])]
for index, ax in enumerate(axes):
    draw_real_panel(ax, index)
fig.text(.28,.055,'(a) ADS Construction',ha='center',fontsize=7)
fig.text(.79,.055,'(b) Proof Path Generation & Updates',ha='center',fontsize=6)
show_figure(fig, 'figure4_real_datasets')
```

## Figure 5 — Scalability Analysis: Impact of Dataset Size $N$

类别数为10。左图为构建时间，右图为平均单次操作耗时；保留全部五个规模的数据点。


```python
def plot_ht_scalability(group):
    compact_style()
    is_scale = group == 'scale'
    xs = [10000,100000,1000000,2000000,5000000] if is_scale else [1,10,100,1000,5000]
    key = 'n' if is_scale else 'm'
    xlabel = 'Data items, $N$' if is_scale else 'Categories, $m$'
    def value(op, x):
        return ht_value(group, op, **{key:x})
    fig = plt.figure(figsize=(3.53, 1.98))
    axes = [fig.add_axes([.105,.29,.36,.54]), fig.add_axes([.625,.29,.36,.54])]
    axes[0].plot(xs,[value('build',x)/1e9 for x in xs],marker='o',
                 markerfacecolor='white',markeredgewidth=.65,color=OP_COLORS[0])
    axes[0].set_ylabel('Time (s)')
    if is_scale:
        axes[0].set_yscale('log')
        axes[0].set_ylim(.0019,1.8)
        axes[0].yaxis.set_major_locator(FixedLocator([.01,.1,1]))
    else:
        axes[0].set_ylim(.15,2.15)
        axes[0].set_yticks([.5,1,1.5,2])
    for op,c,m,ls,label in zip(OPS,OP_COLORS,OP_MARKERS,OP_STYLES,OP_LABELS):
        axes[1].plot(xs,[value(op,x)/1e3 for x in xs],marker=m,
                     markerfacecolor='white',markeredgewidth=.65,
                     color=c,linestyle=ls,label=label)
    axes[1].set_ylabel('Time (µs)')
    axes[1].set_yscale('log')
    axes[1].set_ylim((.2,12) if is_scale else (.4,420))
    axes[1].yaxis.set_major_locator(FixedLocator([1,10] if is_scale else [1,10,100]))
    for ax in axes:
        ax.set_xscale('log')
        ax.set_xlabel(xlabel)
        ticks = [10000,100000,1000000,5000000] if is_scale else xs
        labels = ['10k','100k','1M','5M'] if is_scale else ['1','10','100','1k','5k']
        ax.set_xticks(ticks,labels)
        ax.xaxis.set_minor_locator(NullLocator())
        ax.yaxis.set_minor_locator(NullLocator())
        ax.set_xlim(xs[0]/1.35,xs[-1]*1.35)
        ax.grid(True,alpha=.23,linewidth=.45)
    fig.legend(*axes[1].get_legend_handles_labels(),loc='upper center',
               bbox_to_anchor=(.5,1.025),ncol=2,frameon=False,columnspacing=1.8,
               handlelength=2,handletextpad=.5,labelspacing=.25)
    fig.text(.285,.05,'(a) Construction',ha='center',fontsize=7)
    fig.text(.805,.05,'(b) Runtime Operations',ha='center',fontsize=7)
    return fig

show_figure(plot_ht_scalability('scale'), 'figure5_dataset_size')
```

## 7.6 — Scalability Comparison: Impact of Dataset Size $N$

五种认证结构在相同分类组织方式下的横向对比。五个子图依次为构建、证明生成、插入、修改和删除。

所选 Merkle 库的更新测试未在每轮10秒限时内全部完成计划操作，因此保持省略相应曲线，不将缺失结果当作0。


```python
def comparison_style():
    plt.rcdefaults()
    plt.rcParams.update({
        'font.family':'DejaVu Serif','font.size':9,'axes.titlesize':10,
        'axes.labelsize':9,'legend.fontsize':8,'xtick.labelsize':8,
        'ytick.labelsize':8,'pdf.fonttype':42,'ps.fonttype':42,
        'savefig.dpi':220,'figure.dpi':150,
        'axes.spines.top':False,'axes.spines.right':False,
        'axes.grid':True,'grid.alpha':.2,'lines.linewidth':1.35,'lines.markersize':4})

KINDS = ['ht','merkle','iavl','smt','mpt']
LABEL = {'ht':'HT-Merkle (ours)','merkle':'Merkle','iavl':'IAVL','smt':'SMT','mpt':'MPT'}
COLORS = ['#0072B2','#E69F00','#009E73','#CC79A7','#D55E00']
MARKERS = ['o','s','^','D','v']
STYLES = ['-','--','-.',':',(0,(3,1,1,1))]

def comparison_rows(**filters):
    return [r for r in COMPARISON_DATA if all(r.get(k)==v for k,v in filters.items())]

comparison_style()
fig = plt.figure(figsize=(7.25,4.7),layout='constrained')
grid = fig.add_gridspec(2,6)
axes = [fig.add_subplot(grid[0,:3]), fig.add_subplot(grid[0,3:]),
        fig.add_subplot(grid[1,:2]), fig.add_subplot(grid[1,2:4]),
        fig.add_subplot(grid[1,4:])]
for ax,op,div,title in zip(axes,['build','prove','insert','modify','delete'],
        [1e9,1e3,1e3,1e3,1e3],['(a) Construction','(b) Proof generation',
                               '(c) Insertion','(d) Modification','(e) Deletion']):
    for kind,color,marker,style in zip(KINDS,COLORS,MARKERS,STYLES):
        rs = sorted(comparison_rows(group='scale',kind=kind,operation=op), key=lambda r:r['n'])
        if not rs:
            continue
        assert len({r['n'] for r in rs}) == len(rs)
        ax.plot([r['n'] for r in rs],[r['mean_ns']/div for r in rs],
                color=color,marker=marker,markerfacecolor='none',linestyle=style,label=LABEL[kind])
    ax.set_xscale('log')
    ax.set_yscale('log')
    ax.set_xlabel('Number of data items, $N$')
    ax.set_ylabel('Time (s)' if op=='build' else 'Time (µs)')
    ax.set_title(title)
entries = {}
for ax in axes:
    hs,ls = ax.get_legend_handles_labels()
    for h,l in zip(hs,ls):
        entries.setdefault(l,h)
labels = [LABEL[k] for k in KINDS if LABEL[k] in entries]
fig.legend([entries[l] for l in labels],labels,loc='lower center',ncol=3,
           bbox_to_anchor=(.5,1.01),frameon=False)
show_figure(fig, 'comparison_scale')
```
